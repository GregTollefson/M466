```julia
using Pkg
Pkg.add("CUDA")
```

```julia
using CUDA
using LinearAlgebra
```

```julia
N = 2000

A = rand(Float32, N, N)
B = rand(Float32, N, N)
```

```julia
@time C_cpu = A * B
```

```julia
A_gpu = cu(A)
B_gpu = cu(B)
```

```julia
@time C_gpu = A_gpu * B_gpu
```

```julia
C_gpu_cpu = Array(C_gpu)
```

```julia
maximum(abs.(C_cpu .- C_gpu_cpu))
```

```julia
using Pkg
Pkg.add("BenchmarkTools")
```

```julia
using BenchmarkTools

@benchmark CUDA.@sync $A_gpu * $B_gpu
```

```julia
@benchmark $A * $B
```

```julia
@benchmark $A * $B
@benchmark CUDA.@sync $A_gpu * $B_gpu
```

```julia
cpu_bench = @benchmark $A * $B
```

```julia
gpu_bench = @benchmark CUDA.@sync $A_gpu * $B_gpu
```

```julia
println("CPU:")
display(cpu_bench)

println("GPU:")
display(gpu_bench)
```
